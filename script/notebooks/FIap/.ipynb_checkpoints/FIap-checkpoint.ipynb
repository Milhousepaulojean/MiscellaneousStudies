# Tech Challenge — Impacto do Atraso na Entrega sobre o Faturamento (Olist)

## Pergunta Norteadora
Como o atraso na entrega impacta o faturamento futuro da companhia?

## Decisão Metodológica (Correção Central)
Este notebook corrige um problema estrutural da versão anterior: as tabelas de
`payments`, `order_reviews` e `order_items` possuem granularidade diferente de
`orders` (podem existir múltiplas linhas por pedido). Junções diretas nessas
tabelas duplicavam receita, notas e dias de atraso.

Aqui, definimos duas bases analíticas:
- **df_pedidos**: uma linha por pedido (usada para receita, atraso, satisfação, retenção).
- **df_itens**: uma linha por item (usada para categoria, produto, vendedor).

Todas as métricas financeiras foram recalculadas sobre essa nova estrutura.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configuração visual padrão para todos os gráficos do relatório
sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 100

# CORREÇÃO: centralizamos as importações em uma única célula.
# Isso evita que o notebook dependa de imports espalhados (ex: 'import numpy as np'
# aparecendo no meio do código, como ocorria no notebook original).
```

```python
# 1. Carregando todas as tabelas necessárias de uma vez
orders = pd.read_csv('database/olist_orders_dataset.csv')
payments = pd.read_csv('database/olist_order_payments_dataset.csv')
reviews = pd.read_csv('database/olist_order_reviews_dataset.csv')
customers = pd.read_csv('database/olist_customers_dataset.csv')
items = pd.read_csv('database/olist_order_items_dataset.csv')
products = pd.read_csv('database/olist_products_dataset.csv')
sellers = pd.read_csv('database/olist_sellers_dataset.csv')
category_translation = pd.read_csv('database/product_category_name_translation.csv')

# POR QUE: carregar tudo no início (e não em blocos separados ao longo do
# notebook, como no original) facilita auditoria e evita reprocessamento
# de merges intermediários (df_projeto, df_extra, df_final) que existiam antes.

print("Pedidos:", orders.shape)
print("Pagamentos:", payments.shape)
print("Avaliações:", reviews.shape)
print("Clientes:", customers.shape)
print("Itens:", items.shape)
```

## Etapa 1 — Verificação de Granularidade e Qualidade

Antes de qualquer junção, verificamos se `order_id` é único em cada tabela.
Essa checagem é o que faltava no notebook original e é a causa raiz da
possível duplicação de receita.

```python
def resumo_granularidade(df, nome, chave='order_id'):
    """
    Mostra o total de linhas, quantos order_id únicos existem e quantos
    estão duplicados. Isso revela se a tabela tem granularidade 1:1 ou 1:N
    em relação a pedidos.
    """
    total = len(df)
    unicos = df[chave].nunique()
    duplicados = total - unicos
    print(f"--- {nome} ---")
    print(f"Linhas totais: {total:,}")
    print(f"{chave} únicos: {unicos:,}")
    print(f"Linhas 'extras' (duplicadas em relação a {chave}): {duplicados:,}\n")

resumo_granularidade(orders, 'orders')
resumo_granularidade(payments, 'payments')
resumo_granularidade(reviews, 'order_reviews')
resumo_granularidade(items, 'order_items')

# POR QUE: esta é a evidência objetiva de que orders.merge(payments).merge(reviews)
# no notebook original poderia multiplicar linhas. Se 'payments' ou 'reviews'
# tiverem mais linhas do que order_id únicos, a duplicação é confirmada.
```

## Etapa 2 — Conversão de Datas e Filtro de Pedidos Válidos

Mantemos a lógica original de remover pedidos sem `order_delivered_customer_date`,
mas documentamos essa decisão como uma limitação (pedidos cancelados/não entregues
saem da análise de atraso).


```python
cols_data = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in cols_data:
    orders[col] = pd.to_datetime(orders[col])

# 1. Guardamos o total antes do filtro para documentar a perda de dados
total_pedidos_antes = len(orders)

# 2. Filtramos apenas pedidos com data de entrega registrada
orders_entregues = orders.dropna(subset=['order_delivered_customer_date']).copy()

total_pedidos_depois = len(orders_entregues)
pct_removido = (1 - total_pedidos_depois / total_pedidos_antes) * 100

print(f"Pedidos totais: {total_pedidos_antes:,}")
print(f"Pedidos entregues (usados na análise): {total_pedidos_depois:,}")
print(f"Percentual removido (sem data de entrega): {pct_removido:.2f}%")

# POR QUE: diferente do notebook original, aqui registramos explicitamente
# quantos pedidos foram descartados e por quê. Isso é exigido pela governança
# de dados do Tech Challenge (documentação de qualidade e reprodutibilidade).
```

```python
# Dias de atraso: positivo = atrasado, negativo = entregue antes do prazo
orders_entregues['dias_atraso'] = (
    orders_entregues['order_delivered_customer_date'] -
    orders_entregues['order_estimated_delivery_date']
).dt.days

orders_entregues['foi_atraso'] = orders_entregues['dias_atraso'] > 0

# CORREÇÃO: esse cálculo é feito ANTES de qualquer merge com payments/reviews,
# garantindo que cada order_id tenha exatamente um valor de dias_atraso.
# No notebook original, esse cálculo já vinha de uma base potencialmente duplicada.

print(orders_entregues['dias_atraso'].describe())
```

## Etapa 3 — Consolidação de Pagamentos por Pedido

Um pedido pode ter mais de uma linha em `payments` (ex: pagamento parcial no
cartão + voucher). Agregamos por `order_id` antes de unir com `orders`.

```python
payments_agg = payments.groupby('order_id').agg(
    payment_value=('payment_value', 'sum'),          # valor total pago no pedido
    qtd_pagamentos=('payment_value', 'count'),        # quantos registros de pagamento
    max_installments=('payment_installments', 'max'), # maior número de parcelas
    payment_type_principal=('payment_type', lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
).reset_index()

# Verificação: quantos pedidos tinham mais de um registro de pagamento?
multi_pagamento = (payments_agg['qtd_pagamentos'] > 1).sum()
print(f"Pedidos com mais de um registro de pagamento: {multi_pagamento:,} "
      f"({multi_pagamento / len(payments_agg) * 100:.2f}%)")

# POR QUE: isso evita que o merge direto (como no original) duplique o pedido
# quando há múltiplos métodos de pagamento. Agora 'payment_value' representa
# o valor TOTAL pago no pedido, não um fragmento dele.
```

## Etapa 4 — Consolidação de Avaliações por Pedido

Alguns pedidos podem ter mais de uma avaliação registrada. Definimos a regra:
manter a avaliação mais recente por `order_id` (critério documentado).

```python
reviews['review_creation_date'] = pd.to_datetime(reviews['review_creation_date'])

# Regra definida: em caso de múltiplas avaliações, mantemos a mais recente
reviews_dedup = (
    reviews.sort_values('review_creation_date')
    .drop_duplicates(subset='order_id', keep='last')
    [['order_id', 'review_score', 'review_creation_date']]
)

duplicadas_removidas = len(reviews) - len(reviews_dedup)
print(f"Avaliações duplicadas por pedido removidas: {duplicadas_removidas:,}")

# POR QUE: no notebook original, orders.merge(reviews) podia gerar mais de
# uma linha por pedido se houvesse mais de uma avaliação. Aqui garantimos
# 1 avaliação = 1 pedido, com a regra explícita documentada.
```

## Etapa 5 — Construção de `df_pedidos` (Uma Linha por Pedido)

Esta é a tabela central da análise financeira, de satisfação e de retenção.


```python
df_pedidos = (
    orders_entregues
    .merge(payments_agg, on='order_id', how='left')
    .merge(reviews_dedup, on='order_id', how='left')
    .merge(customers, on='customer_id', how='left')
)

# Validação de cardinalidade: o número de linhas deve ser IGUAL ao número
# de pedidos entregues, pois cada merge foi feito contra tabelas já agregadas.
assert len(df_pedidos) == len(orders_entregues), (
    "ERRO: df_pedidos duplicou pedidos. Verifique agregações de payments/reviews."
)

print(f"df_pedidos: {df_pedidos.shape[0]:,} linhas, "
      f"{df_pedidos['order_id'].nunique():,} order_id únicos")

# POR QUE: o assert é uma rede de segurança que o notebook original não tinha.
# Se qualquer agregação anterior falhar, o notebook para aqui em vez de
# propagar receita duplicada para os gráficos e conclusões.
```

## Etapa 6 — Construção de `df_itens` (Uma Linha por Item)

Usada exclusivamente para análises de categoria, produto e vendedor, evitando
repetir o valor total do pedido em cada item.

```python
df_itens = (
    items
    .merge(products, on='product_id', how='left')
    .merge(category_translation, on='product_category_name', how='left')
    .merge(sellers, on='seller_id', how='left')
    .merge(
        orders_entregues[['order_id', 'customer_id', 'dias_atraso', 'foi_atraso',
                           'order_purchase_timestamp']],
        on='order_id', how='inner'  # inner: mantemos apenas itens de pedidos entregues
    )
    .merge(customers[['customer_id', 'customer_state', 'customer_unique_id']],
           on='customer_id', how='left')
    .merge(reviews_dedup[['order_id', 'review_score']], on='order_id', how='left')
)

# Receita comercial do item = preço + frete (NÃO o payment_value do pedido)
df_itens['receita_item'] = df_itens['price'].fillna(0) + df_itens['freight_value'].fillna(0)

print(f"df_itens: {df_itens.shape[0]:,} linhas "
      f"(esperado: maior que o número de pedidos, pois há múltiplos itens por pedido)")

# POR QUE: aqui usamos 'price + freight_value' (nível de item) em vez de
# 'payment_value' (nível de pedido). No notebook original, 'payment_value'
# do pedido era repetido para cada item/categoria, inflando a receita por
# categoria em pedidos com múltiplos itens.
```

## 1. Análise Descritiva — Satisfação vs. Atraso

Adicionamos mediana e percentual de notas baixas, não apenas a média, pois a
distribuição de `review_score` não é necessariamente simétrica.

```python
satisfacao = (
    df_pedidos.dropna(subset=['review_score'])
    .groupby('foi_atraso')
    .agg(
        nota_media=('review_score', 'mean'),
        nota_mediana=('review_score', 'median'),
        pct_nota_baixa=('review_score', lambda x: (x <= 2).mean() * 100),
        n_pedidos=('order_id', 'nunique')
    )
    .reset_index()
)
satisfacao['status_entrega'] = satisfacao['foi_atraso'].map({True: 'Com Atraso', False: 'No Prazo'})

print(satisfacao)

# POR QUE: a média sozinha (como no original: 4.28 -> 2.26) pode ocultar
# a forma da distribuição. Mostrar a mediana e o % de notas <=2 dá mais
# robustez à afirmação de que o atraso reduz a satisfação.
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

cores_alerta = {'No Prazo': 'green', 'Com Atraso': 'red'}
sns.barplot(data=satisfacao, x='status_entrega', y='nota_media',
            palette=cores_alerta, ax=axes[0])
axes[0].set_title('Nota Média por Status de Entrega')
axes[0].set_ylabel('Nota Média (review_score)')

sns.barplot(data=satisfacao, x='status_entrega', y='pct_nota_baixa',
            palette=cores_alerta, ax=axes[1])
axes[1].set_title('% de Notas Baixas (≤2) por Status de Entrega')
axes[1].set_ylabel('% de Notas ≤ 2')

plt.tight_layout()
plt.show()
```

## 2. Outliers em `dias_atraso`

Antes de calcular qualquer média de atraso, verificamos a presença de outliers
usando a regra do IQR (conforme Aula 6 de Estatística Essencial).

```python
Q1 = df_pedidos['dias_atraso'].quantile(0.25)
Q3 = df_pedidos['dias_atraso'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers = df_pedidos[
    (df_pedidos['dias_atraso'] < limite_inferior) |
    (df_pedidos['dias_atraso'] > limite_superior)
]

print(f"Q1={Q1:.1f} | Q3={Q3:.1f} | IQR={IQR:.1f}")
print(f"Limite inferior: {limite_inferior:.1f} | Limite superior: {limite_superior:.1f}")
print(f"Total de outliers em dias_atraso: {len(outliers):,} "
      f"({len(outliers)/len(df_pedidos)*100:.2f}% dos pedidos)")

plt.figure(figsize=(8, 4))
sns.boxplot(x=df_pedidos['dias_atraso'], color='tomato')
plt.title('Boxplot — Dias de Atraso (detecção de outliers)')
plt.xlabel('Dias de atraso (negativo = entregue antes do prazo)')
plt.show()

# POR QUE: o notebook original nunca verificou outliers em dias_atraso.
# Pedidos com atrasos extremos (ex: 100+ dias) podem distorcer médias de
# tempo de transporte e correlações. Aqui isso é medido e reportado.
```

## 3. Medidas de Dispersão — Tempo de Transporte por Status

Calculamos desvio-padrão e coeficiente de variação (CV) para avaliar não só
o tempo médio de transporte, mas sua estabilidade/previsibilidade.


```python
df_pedidos['tempo_postagem'] = (
    df_pedidos['order_delivered_carrier_date'] - df_pedidos['order_approved_at']
).dt.days

df_pedidos['tempo_transporte'] = (
    df_pedidos['order_delivered_customer_date'] - df_pedidos['order_delivered_carrier_date']
).dt.days

dispersao_transporte = df_pedidos.groupby('foi_atraso')['tempo_transporte'].agg(
    media='mean',
    mediana='median',
    desvio_padrao='std'
).reset_index()

dispersao_transporte['cv_%'] = (
    dispersao_transporte['desvio_padrao'] / dispersao_transporte['media'] * 100
)

print(dispersao_transporte)

# POR QUE: o notebook original reportava apenas a média de tempo_transporte
# (27.1 dias). O desvio-padrão e o CV mostram se esse número é estável ou
# se existe grande variabilidade — informação essencial (Aula 5) para saber
# se a média é confiável como resumo.
```

## 4. Correlação — Cuidado na Interpretação

Reproduzimos a análise de correlação, mas com linguagem associativa, não causal.

```python
corr_frete = df_pedidos[['freight_value', 'dias_atraso']].corr().iloc[0, 1] \
    if 'freight_value' in df_pedidos.columns else np.nan

# Peso médio vem de df_itens (nível de item), agregado por pedido
peso_por_pedido = df_itens.groupby('order_id')['product_weight_g'].mean().reset_index()
df_pedidos_peso = df_pedidos.merge(peso_por_pedido, on='order_id', how='left')
corr_peso = df_pedidos_peso[['product_weight_g', 'dias_atraso']].corr().iloc[0, 1]

print(f"Correlação frete x atraso: {corr_frete:.2f}")
print(f"Correlação peso x atraso: {corr_peso:.2f}")
print("\nInterpretação: correlações próximas de zero indicam AUSÊNCIA de relação "
      "linear detectável, não uma prova definitiva de que a malha logística é "
      "'a' causa do atraso. Outras variáveis (região, sazonalidade, transportadora) "
      "não foram controladas nesta análise.")

# POR QUE: no original, a frase "prova ineficiência sistêmica" é uma conclusão
# causal forte baseada apenas em correlação fraca. Aqui a redação é ajustada
# para refletir o que os dados de fato permitem afirmar.
```

## 5. Receita em Risco — Recalculada sem Duplicação

Usamos `df_pedidos` (uma linha por pedido) para este cálculo, eliminando o
risco de multiplicação por itens ou pagamentos.

```python
filtro_risco = (df_pedidos['foi_atraso'] == True) & (df_pedidos['review_score'] <= 2)

receita_em_risco = df_pedidos.loc[filtro_risco, 'payment_value'].sum()
pedidos_em_risco = df_pedidos.loc[filtro_risco, 'order_id'].nunique()
receita_total = df_pedidos['payment_value'].sum()
pct_receita_risco = receita_em_risco / receita_total * 100

print(f"Receita total (todos os pedidos entregues): R$ {receita_total:,.2f}")
print(f"Receita em risco (atraso + nota <=2): R$ {receita_em_risco:,.2f}")
print(f"Pedidos em risco: {pedidos_em_risco:,}")
print(f"% da receita total em risco: {pct_receita_risco:.2f}%")

# POR QUE: este número substitui o R$ 723.718,60 do notebook original.
# Ele pode ser igual, maior ou menor — o ponto é que agora é METODOLOGICAMENTE
# confiável, pois vem de uma tabela sem duplicação por pagamento ou avaliação.
```

```python
plt.figure(figsize=(6, 4))
plt.text(0.5, 0.6, f'R$ {receita_em_risco:,.2f}', fontsize=26, color='red',
         ha='center', fontweight='bold')
plt.text(0.5, 0.35, f'{pct_receita_risco:.1f}% da receita total\n'
                     f'({pedidos_em_risco:,} pedidos)',
         fontsize=13, ha='center')
plt.axis('off')
plt.title('Receita em Risco (Recalculada — base sem duplicação)')
plt.show()
```

## 6. Receita em Risco por Estado — Absoluta e Relativa

Mantemos as duas visões (volume e taxa relativa), mas ambas calculadas sobre
`df_pedidos`.

```python
risco_por_estado = (
    df_pedidos.loc[filtro_risco]
    .groupby('customer_state', as_index=False)
    .agg(receita_em_risco=('payment_value', 'sum'),
         pedidos_em_risco=('order_id', 'nunique'))
)

receita_total_uf = (
    df_pedidos.groupby('customer_state', as_index=False)['payment_value']
    .sum().rename(columns={'payment_value': 'receita_total'})
)

comparativo_uf = receita_total_uf.merge(risco_por_estado, on='customer_state', how='left').fillna(0)
comparativo_uf['taxa_risco_%'] = comparativo_uf['receita_em_risco'] / comparativo_uf['receita_total'] * 100

print("Top 10 — Volume absoluto de risco:")
print(comparativo_uf.sort_values('receita_em_risco', ascending=False).head(10))

print("\nTop 10 — Taxa relativa de risco (ineficiência proporcional):")
print(comparativo_uf.sort_values('taxa_risco_%', ascending=False).head(10))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

top_volume = comparativo_uf.sort_values('receita_em_risco', ascending=False).head(10)
sns.barplot(data=top_volume, x='receita_em_risco', y='customer_state',
            palette='Reds_r', ax=axes[0])
axes[0].set_title('Top 10 Estados — Volume de Receita em Risco')
axes[0].set_xlabel('R$ em risco')

top_taxa = comparativo_uf.sort_values('taxa_risco_%', ascending=False).head(10)
sns.barplot(data=top_taxa, x='taxa_risco_%', y='customer_state',
            palette='Oranges_r', ax=axes[1])
axes[1].set_title('Top 10 Estados — Taxa Relativa de Risco (%)')
axes[1].set_xlabel('% da receita local em risco')

plt.tight_layout()
plt.show()
```

## 7. Receita em Risco por Categoria — Usando `df_itens`

Diferente do original, aqui a receita por categoria é calculada com
`price + freight_value` do item, não com `payment_value` do pedido inteiro.

```python
filtro_risco_itens = (df_itens['foi_atraso'] == True) & (df_itens['review_score'] <= 2)

risco_por_categoria = (
    df_itens.loc[filtro_risco_itens]
    .groupby('product_category_name_english', as_index=False)
    .agg(receita_em_risco=('receita_item', 'sum'),
         itens_em_risco=('order_item_id', 'count'),
         pedidos_em_risco=('order_id', 'nunique'))
    .sort_values('receita_em_risco', ascending=False)
)

print("Top 5 categorias em risco (base: preço + frete do item):")
print(risco_por_categoria.head(5))

plt.figure(figsize=(9, 5))
sns.barplot(data=risco_por_categoria.head(5), x='receita_em_risco',
            y='product_category_name_english', palette='Reds_r')
plt.title('Top 5 Categorias — Receita em Risco (nível de item)')
plt.xlabel('R$ em risco (price + freight_value)')
plt.tight_layout()
plt.show()

# POR QUE: usar 'receita_item' evita que o payment_value total do pedido seja
# contado uma vez para CADA item/categoria, o que inflava artificialmente
# categorias em pedidos com múltiplos produtos no notebook original.
```

## 8. Perfil de Cliente — Novos vs. Recorrentes

Usamos `customer_unique_id` (identifica a pessoa, não o cadastro por pedido)
para medir corretamente a recorrência.

```python
freq_cliente = (
    df_pedidos.groupby('customer_unique_id')['order_id']
    .nunique().reset_index(name='total_pedidos')
)

df_pedidos = df_pedidos.merge(freq_cliente, on='customer_unique_id', how='left')
df_pedidos['perfil_cliente'] = np.where(
    df_pedidos['total_pedidos'] > 1, 'VIP (Recorrente)', 'Novo Cliente'
)

risco_perfil = (
    df_pedidos.loc[filtro_risco]
    .groupby('perfil_cliente', as_index=False)['payment_value']
    .sum()
)

print(risco_perfil)

# LIMITAÇÃO DOCUMENTADA: "total_pedidos > 1" indica que o cliente comprou mais
# de uma vez ATÉ O FIM DO DATASET (2016-2018), não necessariamente APÓS o
# pedido atrasado. Isto NÃO comprova causalidade entre atraso e não-recompra;
# é uma correlação a ser aprofundada com análise de coorte (compra antes/depois).
```

```python
plt.figure(figsize=(7, 5))
grafico = sns.barplot(data=risco_perfil, x='perfil_cliente', y='payment_value',
                       palette='OrRd_r')
plt.title('Risco Financeiro por Perfil de Cliente (Recalculado)')
plt.ylabel('R$ em risco')

for p in grafico.patches:
    grafico.annotate(f'R$ {p.get_height():,.0f}',
                      (p.get_x() + p.get_width()/2, p.get_height()),
                      ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```

## 9. Meta de Entrega — Uso de Mediana em vez de Média

O benchmark de "dias antes do prazo para nota 5" usa a mediana, pois a
distribuição de dias_atraso tem outliers (Etapa 2) que distorceriam a média.
