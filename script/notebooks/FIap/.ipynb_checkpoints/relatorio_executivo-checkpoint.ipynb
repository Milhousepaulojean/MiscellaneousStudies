# Relatório Executivo — Impacto da Eficiência Logística no Faturamento (Olist)

Pergunta Norteadora: **Como a eficiência logística impacta o faturamento dos
e-commerces brasileiros?**

Este script gera todos os gráficos e indicadores executivos derivados do
notebook `Ps_Tech2.ipynb`, cobrindo: atraso na entrega, satisfação do
cliente, receita em risco, gargalos logísticos, sazonalidade, geolocalização
e comportamento de clientes (retenção).

```
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

# Configuração visual padrão
sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 100
os.makedirs("reports", exist_ok=True)

CORES = sns.color_palette("tab10")


def _show_image(path):
    """Exibe a imagem salva (compatível com execução em notebook ou script)."""
    try:
        display(Image(filename=path))
    except Exception:
        print(f"Gráfico salvo em: {path}")
```

---
## 1. Carregamento e Preparação dos Dados

Unificação das tabelas principais (`orders`, `payments`, `reviews`) e
cálculo das colunas de atraso, conforme o notebook original.

```
orders = pd.read_csv("database/olist_orders_dataset.csv")
payments = pd.read_csv("database/olist_order_payments_dataset.csv")
reviews = pd.read_csv("database/olist_order_reviews_dataset.csv")
customers = pd.read_csv("database/olist_customers_dataset.csv")
products = pd.read_csv("database/olist_products_dataset.csv")
items = pd.read_csv("database/olist_order_items_dataset.csv")
sellers = pd.read_csv("database/olist_sellers_dataset.csv")
geo = pd.read_csv("database/olist_geolocation_dataset.csv")

# União principal: orders + payments + reviews
df_projeto = orders.merge(payments, on="order_id").merge(reviews, on="order_id")

cols_data = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]
for col in cols_data:
    df_projeto[col] = pd.to_datetime(df_projeto[col])

# Removendo pedidos sem data de entrega (cancelados/em transporte)
df_projeto = df_projeto.dropna(subset=["order_delivered_customer_date"])

# Cálculo de atraso
df_projeto["dias_atraso"] = (
    df_projeto["order_delivered_customer_date"]
    - df_projeto["order_estimated_delivery_date"]
).dt.days
df_projeto["foi_atraso"] = df_projeto["dias_atraso"] > 0

print(f"Total de registros na base unificada: {df_projeto.shape}")
```

---
## 2. Impacto do Atraso na Satisfação do Cliente

**Insight:** a nota média cai de 4,28 (No Prazo) para 2,26 (Com Atraso),
uma redução de quase 50% na satisfação.

```
analise_satisfacao = (
    df_projeto.groupby("foi_atraso")["review_score"].mean().reset_index()
)
analise_satisfacao["foi_atraso"] = analise_satisfacao["foi_atraso"].map(
    {False: "No Prazo", True: "Com Atraso"}
)

receita_em_risco = df_projeto.loc[
    (df_projeto["foi_atraso"] == True) & (df_projeto["review_score"] <= 2),
    "payment_value",
].sum()

paleta_cores = {"No Prazo": "green", "Com Atraso": "red"}

plt.figure(figsize=(10, 6))
grafico = sns.barplot(
    data=analise_satisfacao, x="foi_atraso", y="review_score", palette=paleta_cores
)
plt.title(
    "Impacto do Atraso na Satisfação do Cliente",
    fontsize=16,
    fontweight="bold",
    pad=20,
)
plt.xlabel("Status da Entrega", fontsize=12)
plt.ylabel("Média da Nota (review_score)", fontsize=12)
plt.ylim(0, 5)

plt.text(
    0.5,
    4.5,
    f"RECEITA EM RISCO:\nR$ {receita_em_risco:,.2f}",
    ha="center",
    va="center",
    fontsize=13,
    color="darkred",
    fontweight="bold",
    bbox=dict(facecolor="white", edgecolor="darkred", boxstyle="round,pad=0.5"),
)

for p in grafico.patches:
    grafico.annotate(
        f"{p.get_height():.2f}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 9),
        textcoords="offset points",
        fontweight="bold",
    )

plt.tight_layout()
plt.savefig("reports/satisfacao_atraso.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/satisfacao_atraso.png")
```

---
## 3. Receita em Risco por Estado

**Insight:** SP e RJ concentram a maior parte do risco financeiro.

```
df_extra = items.merge(products, on="product_id")
df_final = df_projeto.merge(customers, on="customer_id").merge(df_extra, on="order_id")

risco_por_estado = (
    df_final[(df_final["foi_atraso"] == True) & (df_final["review_score"] <= 2)]
    .groupby("customer_state")["payment_value"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=risco_por_estado.head(10),
    x="payment_value",
    y="customer_state",
    palette="Reds_r",
)
plt.title(
    "Top 10 Estados: Volume de Receita em Risco por Atraso",
    fontsize=15,
    fontweight="bold",
)
plt.xlabel("Volume Financeiro em Risco (R$)", fontsize=12)
plt.ylabel("Estado (UF)", fontsize=12)

for i, v in enumerate(risco_por_estado.head(10)["payment_value"]):
    plt.text(v + 1000, i, f"R$ {v:,.0f}", va="center", fontsize=10, fontweight="bold")

plt.tight_layout()
plt.savefig("reports/risco_por_estado.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/risco_por_estado.png")
```

## 4. Gargalo Logístico: Vendedor vs. Transportadora

**Insight:** o problema é logístico, não operacional do vendedor.

```
df_final["tempo_postagem"] = (
    pd.to_datetime(df_final["order_delivered_carrier_date"])
    - df_final["order_purchase_timestamp"]
).dt.days
df_final["tempo_transporte"] = (
    df_final["order_delivered_customer_date"]
    - pd.to_datetime(df_final["order_delivered_carrier_date"])
).dt.days

atrasados = df_final[df_final["foi_atraso"] == True]

dados_comparativos = pd.DataFrame(
    {
        "Responsável": ["Vendedor", "Transportadora"],
        "Média de Dias (Atrasados)": [
            atrasados["tempo_postagem"].mean(),
            atrasados["tempo_transporte"].mean(),
        ],
    }
)

plt.figure(figsize=(10, 6))
grafico_gargalo = sns.barplot(
    data=dados_comparativos,
    x="Responsável",
    y="Média de Dias (Atrasados)",
    palette=["#FFD700", "#D90429"],
)
plt.title(
    "Causa Raiz do Atraso: Onde o Produto Passa Mais Tempo?",
    fontsize=15,
    fontweight="bold",
)
plt.ylabel("Média de Dias")

for p in grafico_gargalo.patches:
    grafico_gargalo.annotate(
        f"{p.get_height():.1f} dias",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 9),
        textcoords="offset points",
        fontweight="bold",
    )

plt.tight_layout()
plt.savefig("reports/gargalo_logistico.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/gargalo_logistico.png")
```

## 5. Sazonalidade do Risco Financeiro

**Insight:** o risco explode na Black Friday e festas de fim de ano
(Nov/2017 a Fev/2018).

```
df_final["mes_compra"] = df_final["order_purchase_timestamp"].dt.to_period("M").astype(str)

evolucao_risco = (
    df_final[(df_final["foi_atraso"] == True) & (df_final["review_score"] <= 2)]
    .groupby("mes_compra")["payment_value"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(14, 6))
sns.lineplot(
    data=evolucao_risco,
    x="mes_compra",
    y="payment_value",
    marker="o",
    color="#D90429",
    linewidth=2.5,
)
plt.title(
    "Evolução Mensal da Receita em Risco (Atrasos Críticos)",
    fontsize=15,
    fontweight="bold",
)
plt.xlabel("Mês da Compra")
plt.ylabel("Volume Financeiro em Risco (R$)")
plt.xticks(rotation=45)
plt.axvspan("2017-11", "2018-02", color="red", alpha=0.1, label="Pico Sazonal (Black Friday/Natal)")
plt.legend()
plt.tight_layout()
plt.savefig("reports/sazonalidade_risco.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/sazonalidade_risco.png")
```

## 6. Frete e Peso vs. Atraso

**Insight:** correlação quase nula (Frete: -0.04 | Peso: 0.03) — o atraso
é uma ineficiência sistêmica, não relacionada ao tipo/custo da carga.

```
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(
    data=df_final[df_final["foi_atraso"] == True],
    x="product_weight_g",
    y="dias_atraso",
    alpha=0.3,
    color="gray",
)
corr_peso = df_final["product_weight_g"].corr(df_final["dias_atraso"])
plt.title(f"Peso da Encomenda vs. Atraso\n(Correlação: {corr_peso:.2f})", fontsize=11)
plt.xlabel("Peso (gramas)")
plt.ylabel("Dias de Atraso")

plt.subplot(1, 2, 2)
sns.scatterplot(
    data=df_final[df_final["foi_atraso"] == True],
    x="freight_value",
    y="dias_atraso",
    alpha=0.3,
    color="gray",
)
corr_frete = df_final["freight_value"].corr(df_final["dias_atraso"])
plt.title(f"Valor do Frete vs. Atraso\n(Correlação: {corr_frete:.2f})", fontsize=11)
plt.xlabel("Valor do Frete (R$)")

plt.tight_layout()
plt.savefig("reports/frete_peso_atraso.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/frete_peso_atraso.png")
```

## 7. Distância entre Vendedor e Comprador vs. Atraso

**Insight:** correlação de -0.08 — a distância física não é o fator
determinante do atraso; o problema é a gestão da malha logística.

```
geo_reduced = (
    geo.groupby("geolocation_zip_code_prefix")[["geolocation_lat", "geolocation_lng"]]
    .mean()
    .reset_index()
)

df_final_with_sellers = df_final.merge(sellers, on="seller_id", how="left")

df_geo = df_final_with_sellers.merge(
    geo_reduced,
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left",
).rename(columns={"geolocation_lat": "lat_cust", "geolocation_lng": "lng_cust"})

df_geo = df_geo.merge(
    geo_reduced,
    left_on="seller_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left",
).rename(columns={"geolocation_lat": "lat_seller", "geolocation_lng": "lng_seller"})

df_geo = df_geo.dropna(subset=["lat_cust", "lng_cust", "lat_seller", "lng_seller"])


def haversine(lat1, lon1, lat2, lon2):
    r = 6371
    phi1, phi2 = np.radians(lat1), np.radians(lat2)
    delta_phi = np.radians(lat2 - lat1)
    delta_lambda = np.radians(lon2 - lon1)
    a = np.sin(delta_phi / 2) ** 2 + np.cos(phi1) * np.cos(phi2) * np.sin(delta_lambda / 2) ** 2
    return r * 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))


df_geo["distancia_km"] = haversine(
    df_geo["lat_cust"], df_geo["lng_cust"], df_geo["lat_seller"], df_geo["lng_seller"]
)

correlacao_dist = df_geo["distancia_km"].corr(df_geo["dias_atraso"])
print(f"Correlação entre Distância e Atraso: {correlacao_dist:.2f}")

max_dist = df_geo["distancia_km"].max()
bins = [0, 250, 500, 1000, 2000, max_dist if max_dist > 2000 else 2001]
df_geo["faixa_distancia"] = pd.cut(
    df_geo["distancia_km"],
    bins=bins,
    labels=["Até 250km", "250-500km", "500-1000km", "1000-2000km", "Acima 2000km"],
    right=True,
    include_lowest=True,
)

analise_distancia = (
    df_geo.groupby("faixa_distancia", observed=False)
    .agg({"foi_atraso": "mean", "review_score": "mean", "tempo_transporte": "mean"})
    .reset_index()
)
analise_distancia["taxa_atraso_%"] = analise_distancia["foi_atraso"] * 100

fig, ax1 = plt.subplots(figsize=(14, 7))
sns.barplot(
    data=analise_distancia,
    x="faixa_distancia",
    y="tempo_transporte",
    ax=ax1,
    palette="Blues_d",
    alpha=0.7,
)
ax1.set_ylabel("Tempo Médio de Transporte (Dias)", fontsize=12, fontweight="bold", color="navy")
ax1.set_xlabel("Faixa de Distância entre Vendedor e Comprador", fontsize=12)

ax2 = ax1.twinx()
sns.lineplot(
    data=analise_distancia,
    x="faixa_distancia",
    y="taxa_atraso_%",
    ax=ax2,
    marker="o",
    color="red",
    linewidth=3,
    label="Taxa de Atraso (%)",
)
ax2.set_ylabel("Taxa de Atraso (%)", fontsize=12, fontweight="bold", color="red")
ax2.grid(False)

plt.title("DIAGNÓSTICO: A distância justifica o atraso?", fontsize=16, fontweight="bold", pad=25)

for p in ax1.patches:
    ax1.annotate(
        f"{p.get_height():.1f} d",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 9),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold",
    )

for x, y in zip(analise_distancia["faixa_distancia"], analise_distancia["taxa_atraso_%"]):
    ax2.annotate(
        f"{y:.1f}%",
        (x, y),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center",
        fontsize=10,
        color="red",
        fontweight="bold",
    )

fig.tight_layout()
plt.savefig("reports/distancia_atraso.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/distancia_atraso.png")
```

## 8. Distribuição Mensal de Satisfação (2016-2018)

```
dist_percentual = pd.crosstab(df_final["mes_compra"], df_final["review_score"], normalize="index") * 100
cores_satisfacao = ["#D90429", "#F48C06", "#FFBA08", "#80B918", "#2D6A4F"]

plt.figure(figsize=(18, 8))
dist_percentual.plot(kind="bar", stacked=True, color=cores_satisfacao, ax=plt.gca(), width=0.85, alpha=0.9)
plt.title(
    "DISTRIBUIÇÃO MENSAL DE SATISFAÇÃO (2016-2018)\nImpacto Visual da Ineficiência Logística na Experiência do Cliente",
    fontsize=18,
    fontweight="bold",
    pad=25,
)
plt.xlabel("Meses da Operação", fontsize=12)
plt.ylabel("Percentual das Avaliações (%)", fontsize=12)
plt.legend(title="Review Score", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 100)
sns.despine()
plt.tight_layout()
plt.savefig("reports/distribuicao_satisfacao_mensal.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/distribuicao_satisfacao_mensal.png")
```

## 9. Risco Financeiro por Perfil de Cliente

**Insight:** risco concentrado em novos clientes — destruindo o LTV logo
na primeira compra.

```
frequencia_cliente = df_final.groupby("customer_unique_id")["order_id"].nunique().reset_index()
frequencia_cliente.columns = ["customer_unique_id", "total_pedidos"]

df_final = df_final.merge(frequencia_cliente, on="customer_unique_id")
df_final["perfil_cliente"] = df_final["total_pedidos"].apply(
    lambda x: "VIP (Recorrente)" if x > 1 else "Novo Cliente"
)

risco_perfil = (
    df_final[(df_final["foi_atraso"] == True) & (df_final["review_score"] <= 2)]
    .groupby("perfil_cliente")["payment_value"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 6))
grafico_perfil = sns.barplot(data=risco_perfil, x="perfil_cliente", y="payment_value", palette="OrRd_r")
plt.title("Risco Financeiro: Novos Clientes vs. Clientes VIP", fontsize=15, fontweight="bold")
plt.ylabel("Volume Financeiro em Risco (R$)")
plt.xlabel("Perfil do Comprador")

for p in grafico_perfil.patches:
    grafico_perfil.annotate(
        f"R$ {p.get_height():,.2f}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 9),
        textcoords="offset points",
        fontweight="bold",
    )

plt.tight_layout()
plt.savefig("reports/risco_perfil_cliente.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/risco_perfil_cliente.png")
```

## 10. Satisfação na Primeira Compra vs. Taxa de Retorno

```
df_sorted = df_final.sort_values(["customer_unique_id", "order_purchase_timestamp"])
primeiro_pedido = df_sorted.groupby("customer_unique_id").first().reset_index()

contagem_pedidos = df_final.groupby("customer_unique_id")["order_id"].nunique().reset_index()
contagem_pedidos.columns = ["customer_unique_id", "total_pedidos"]

df_retencao = primeiro_pedido[["customer_unique_id", "review_score"]].merge(
    contagem_pedidos, on="customer_unique_id"
)
df_retencao["voltou_a_comprar"] = df_retencao["total_pedidos"] > 1


def categorizar(nota):
    if nota <= 3:
        return "1-3 (Mal)"
    if nota == 4:
        return "4 (Bom)"
    return "5 (Ótimo)"


df_retencao["satisfacao_inicial"] = df_retencao["review_score"].apply(categorizar)

taxa_retorno = (
    df_retencao.groupby("satisfacao_inicial")["voltou_a_comprar"].mean() * 100
).reset_index()
taxa_retorno.columns = ["Categoria", "Taxa_de_Retorno_%"]

plt.figure(figsize=(10, 6))
paleta = ["#D90429", "#FFBA08", "#2D6A4F"]
grafico = sns.barplot(data=taxa_retorno, x="Categoria", y="Taxa_de_Retorno_%", palette=paleta)
plt.title(
    "FIDELIDADE EM RISCO: Clientes insatisfeitos voltam a comprar?",
    fontsize=15,
    fontweight="bold",
    pad=20,
)
plt.ylabel("Taxa de Retorno (%)", fontsize=12)
plt.xlabel("Nota da 1ª Experiência", fontsize=12)
plt.ylim(0, taxa_retorno["Taxa_de_Retorno_%"].max() * 1.2)

for p in grafico.patches:
    grafico.annotate(
        f"{p.get_height():.2f}%",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 9),
        textcoords="offset points",
        fontweight="bold",
    )

plt.tight_layout()
plt.savefig("reports/retorno_por_satisfacao.png", dpi=150, bbox_inches="tight", facecolor="white")
plt.close()
_show_image("reports/retorno_por_satisfacao.png")
```

## Recomendações Executivas

| Área | Recomendação |
|------|---------------|
| **Logística** | Investigar a etapa de transporte (27,1 dias) como principal gargalo, não a postagem do vendedor (5,2 dias). |
| **Sazonalidade** | Reforçar capacidade logística entre Nov/2017 e Fev/2018 (Black Friday/Natal). |
| **Retenção** | Priorizar a experiência da primeira compra — novos clientes concentram o maior volume de receita em risco. |
| **Geografia** | Focar ações corretivas em SP e RJ, que concentram a maior parte do risco financeiro absoluto. |
| **Precificação/Frete** | Não usar peso ou valor do frete como preditores de atraso — a causa é a gestão do processo, não a carga. |
